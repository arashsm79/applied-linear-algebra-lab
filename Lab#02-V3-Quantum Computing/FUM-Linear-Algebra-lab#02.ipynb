# 1. The Qubit: Superposition, Measurement, and Normalization

In classical computing, a bit is a switch: it is either **0** or **1**.

In quantum computing, we use a **qubit**. We represent the two basic states using column vectors, a notation often called "Dirac" or "ket" notation:

$$ |0\rangle = \begin{bmatrix} 1 \\ 0 \end{bmatrix} \quad \text{and} \quad |1\rangle = \begin{bmatrix} 0 \\ 1 \end{bmatrix} $$

A qubit can exist in a state that is a **linear combination** of these two. This is called **superposition**. Mathematically, a qubit state $|\psi\rangle$ is a vector:

$$ |\psi\rangle = \alpha|0\rangle + \beta|1\rangle = \begin{bmatrix} \alpha \\ \beta \end{bmatrix} $$

Here, $\alpha$ and $\beta$ are complex numbers called **amplitudes**.

### Measurement and Probability

Unlike a classical bit, we cannot "see" the superposition. When we **measure** a qubit, it "collapses" to either state $|0\rangle$ or $|1\rangle$. The likelihood of this happening depends on the amplitudes:

*   Probability of measuring **0**: $P(0) = |\alpha|^2$
*   Probability of measuring **1**: $P(1) = |\beta|^2$


## The Qubit: From Vector to Unit Vector

As we've seen, a qubit's state $|\psi\rangle$ can be represented as a 2D column vector with complex coefficients $\alpha$ and $\beta$:

$$ |\psi\rangle = \begin{bmatrix} \alpha \\ \beta \end{bmatrix} $$

A fundamental rule in quantum mechanics is that a state vector must have a **norm (or length) of 1**. This is known as the **normalization condition**.

$$ \| |\psi\rangle \| = \sqrt{|\alpha|^2 + |\beta|^2} = 1 $$

Squaring both sides gives the familiar probability rule: $|\alpha|^2 + |\beta|^2 = 1$. Only **unit vectors** are valid quantum states, because they ensure the probabilities of measurement outcomes sum to 100%.

This leads to a three-step process:
1.  **Verify**: Is our vector a valid state (i.e., is it normalized)?
2.  **Correct**: If not, can we normalize it to make it a valid state?
3.  **Use**: Once we have a valid state, we can calculate measurement probabilities.

# 3. Vector Spaces and Linear Combinations

The mathematical environment where quantum states exist and interact is a **vector space**. For students of linear algebra, this is familiar territory.

The specific vector space important for quantum computing is **ℂⁿ**, which is the vector space of "n-tuples" of complex numbers. An "n-tuple" is simply an ordered collection of *n* numbers. We represent these as n-dimensional column vectors.

For example, a state vector `|a⟩` in ℂⁿ would look like this:

$$ |a\rangle = \begin{bmatrix} a_1 \\ a_2 \\ \vdots \\ a_n \end{bmatrix} $$

### Connection to the Qubit

You've already been working in a vector space! The single qubit state $|\psi\rangle = \begin{bmatrix} \alpha \\ \beta \end{bmatrix}$ is a vector in **ℂ²**, the vector space of 2-tuples of complex numbers. The basis vectors $|0\rangle$ and $|1\rangle$ form a basis for this space.

### Operations in a Vector Space

A vector space is defined by two fundamental operations:

**1. Vector Addition:** Two vectors in the same space can be added together component-wise to produce a new vector in that space.

$$ |a\rangle + |b\rangle = \begin{bmatrix} a_1 \\ a_2 \\ \vdots \\ a_n \end{bmatrix} + \begin{bmatrix} b_1 \\ b_2 \\ \vdots \\ b_n \end{bmatrix} = \begin{bmatrix} a_1 + b_1 \\ a_2 + b_2 \\ \vdots \\ a_n + b_n \end{bmatrix} $$

**2. Scalar Multiplication:** Any vector can be multiplied by a scalar (in our case, a complex number), which scales the vector.

$$ c|a\rangle = c \begin{bmatrix} a_1 \\ a_2 \\ \vdots \\ a_n \end{bmatrix} = \begin{bmatrix} ca_1 \\ ca_2 \\ \vdots \\ ca_n \end{bmatrix} $$

These two operations together are what allow us to form **linear combinations**—the very tool we used to define a qubit's superposition.

# 4. Basis and Dimension

A **basis** for a vector space is a set of vectors with two key properties:
1.  **Linearly Independent**: None of the vectors in the set can be written as a linear combination of the others.
2.  **Spanning**: Any vector in the entire space can be created by a linear combination of the basis vectors.

The **dimension** of a vector space is simply the number of vectors in its basis. Since we've established that a single qubit lives in $\mathbb{C}^2$, we know it's a 2-dimensional space, so its basis must contain exactly two vectors.

### The Standard (or Computational) Basis

So far, we have been using the most common basis, known as the **standard basis** or **computational basis**:

$$ |0\rangle = \begin{bmatrix} 1 \\ 0 \end{bmatrix}, \quad |1\rangle = \begin{bmatrix} 0 \\ 1 \end{bmatrix} $$

When we write a qubit state $|\psi\rangle = \begin{bmatrix} \alpha \\ \beta \end{bmatrix}$, the coefficients $\alpha$ and $\beta$ are implicitly the "amounts" of $|0\rangle$ and $|1\rangle$ needed to build $|\psi\rangle$.

### A Different Basis: The Hadamard Basis

A vector space can have many different bases. Another important basis for $\mathbb{C}^2$ is constructed from two vectors often called $|+\rangle$ and $|-\rangle$:

$$ |+\rangle = \frac{1}{\sqrt{2}}\begin{bmatrix} 1 \\ 1 \end{bmatrix}, \quad |-\rangle = \frac{1}{\sqrt{2}}\begin{bmatrix} 1 \\ -1 \end{bmatrix} $$

Just like any qubit state can be written as $|\psi\rangle = \alpha|0\rangle + \beta|1\rangle$, it can *also* be written as a linear combination of $|+\rangle$ and $|-\rangle$:

$$ |\psi\rangle = \gamma|+\rangle + \delta|-\rangle $$

Finding the coefficients $(\gamma, \delta)$ for a new basis is a fundamental operation in linear algebra known as a **change of basis**.

# 5. The Inner Product: Measuring and Projecting

The **inner product** is a generalization of the dot product for complex vector spaces. While a dot product takes two real vectors and gives a real number (a scalar), the inner product takes two complex vectors and returns a complex number.

### "Bra-Ket" Notation

So far, we have used "ket" vectors like $|\psi\rangle$ to represent quantum states as column vectors.

$$ |\psi\rangle = \begin{bmatrix} \alpha \\ \beta \end{bmatrix} $$

For every ket, there is a corresponding "bra" vector, written as $\langle\psi|$. A "bra" is the **conjugate transpose** of its corresponding "ket". The transpose operation turns the column vector into a row vector, and the conjugate operation takes the complex conjugate of each element.

$$ \langle\psi| = (|\psi\rangle)^\dagger = \begin{bmatrix} \alpha^* & \beta^* \end{bmatrix} $$

*(Recall that the complex conjugate of a number $z = x + iy$ is $z^* = x - iy$.)*

### Calculating the Inner Product

The inner product between two vectors, $|u\rangle$ and $|v\rangle$, is written as $\langle u | v \rangle$. This "bra-ket" (bracket) is calculated by multiplying the bra of the first vector with the ket of the second vector:

$$ \langle u|v \rangle = \begin{bmatrix} u_1^* & u_2^* \end{bmatrix} \begin{bmatrix} v_1 \\ v_2 \end{bmatrix} = u_1^*v_1 + u_2^*v_2 $$

### Properties of the Inner Product

1.  **Defining the Norm**: The inner product of a vector with itself, $\langle\psi|\psi\rangle$, gives the square of its norm. Since $|\alpha|^2 + |\beta|^2 = 1$ for a valid qubit, we have:
    $$ \langle\psi|\psi\rangle = 1 $$

2.  **Orthogonality**: Two vectors $|u\rangle$ and $|v\rangle$ are **orthogonal** if their inner product is zero.
    $$ \langle u|v \rangle = 0 $$
    For example, our basis states $|0\rangle$ and $|1\rangle$ are orthogonal:
    $$ \langle 0|1 \rangle = \begin{bmatrix} 1 & 0 \end{bmatrix} \begin{bmatrix} 0 \\ 1 \end{bmatrix} = (1)(0) + (0)(1) = 0 $$
    This is why we call them an **orthonormal basis** (orthogonal and normalized).

# 6. Orthonormality and Creating a Basis

### Orthonormality
We have seen two key properties of vectors:
1.  **Normalization**: A vector is *normalized* if its norm is 1. That is, $\langle\psi|\psi\rangle = 1$.
2.  **Orthogonality**: Two vectors are *orthogonal* if their inner product is 0. That is, $\langle u|v \rangle = 0$.

A set of vectors that satisfies **both** of these properties is called an **orthonormal set**. Specifically, if you have a set of vectors $\{|u_1\rangle, |u_2\rangle, \dots, |u_n\rangle\}$, it is orthonormal if:

$$ \langle u_i | u_j \rangle = \delta_{ij} = \begin{cases} 1 & \text{if } i=j \\ 0 & \text{if } i \neq j \end{cases} $$

The standard basis $\{|0\rangle, |1\rangle\}$ is the simplest example of an orthonormal basis for $\mathbb{C}^2$.

### The Gram-Schmidt Process
What if you have a set of basis vectors that *aren't* orthonormal? The **Gram-Schmidt process** is a powerful and systematic algorithm that takes any linearly independent set of vectors and transforms it into an orthonormal basis. You must be familiar with the consepts in the course materials.

The process is iterative. For each vector in your starting basis $\{|v_1\rangle, |v_2\rangle, \dots, |v_n\rangle\}$, you make it orthogonal to all the *new, orthonormal* vectors you've already created.

Let's build the new orthonormal basis $\{|u_1\rangle, |u_2\rangle, \dots, |u_n\rangle\}$ step-by-step:

1.  **First Vector**: For the first vector, you simply normalize it.
    $$ |u_1\rangle = \frac{|v_1\rangle}{\|v_1\|} $$

2.  **Second Vector**: For the second vector, $|v_2\rangle$, you first subtract its projection onto $|u_1\rangle$. This gives you an intermediate vector, $|w_2\rangle$, that is orthogonal to $|u_1\rangle$.
    $$ |w_2\rangle = |v_2\rangle - \langle u_1 | v_2 \rangle |u_1\rangle $$
    Then you normalize $|w_2\rangle$ to get your second orthonormal vector, $|u_2\rangle$.
    $$ |u_2\rangle = \frac{|w_2\rangle}{\|w_2\|} $$

3.  **k-th Vector**: For any subsequent vector $|v_k\rangle$, you subtract its projection onto *all* the previously computed orthonormal vectors $|u_1\rangle, \dots, |u_{k-1}\rangle$:
    $$ |w_k\rangle = |v_k\rangle - \sum_{j=1}^{k-1} \langle u_j | v_k \rangle |u_j| $$
    And then, as always, you normalize the resulting orthogonal vector:
    $$ |u_k\rangle = \frac{|w_k\rangle}{\|w_k\|} $$

You repeat this process until all vectors from the original basis have been used.

# 7. Quantum Gates: Operators as Matrices

So far, we have focused on describing the state of a qubit as a vector. Now, we will take action. How do we change a qubit from one state to another?

In quantum mechanics, any operation that transforms a quantum state is called an **operator**. In the context of linear algebra, an operator is simply a **matrix**. Applying an operator to a qubit is equivalent to matrix-vector multiplication.

If `A` is an operator (a matrix) and `|ψ⟩` is a qubit state (a vector), the new state `|φ⟩` is found by:

$$ |φ\rangle = A|\psi\rangle $$

These operators are often called **quantum gates**, in analogy to the logic gates in classical computers.

### The Identity Operator

The simplest operator is the **Identity operator (I)**, which leaves a state unchanged. It's the matrix equivalent of doing nothing.

$$ I = \begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix} \implies I|\psi\rangle = \begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix} \begin{bmatrix} \alpha \\ \beta \end{bmatrix} = \begin{bmatrix} \alpha \\ \beta \end{bmatrix} = |\psi\rangle $$

### The Pauli Operators: The Fundamental Gates

A set of three operators of fundamental importance are the **Pauli Operators**, denoted `X`, `Y`, and `Z`.

#### 1. The Pauli-X Gate (The NOT Gate)

The `X` gate is the quantum equivalent of a classical NOT gate. It flips the state $|0\rangle$ to $|1\rangle$ and vice-versa.

*   $X|0\rangle = |1\rangle$
*   $X|1\rangle = |0\rangle$

From these actions, we can derive its matrix form:
$$ X = \begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix} $$

#### 2. The Pauli-Z Gate (The Phase-Flip Gate)

The `Z` gate leaves $|0\rangle$ unchanged but flips the sign (the phase) of $|1\rangle$.

*   $Z|0\rangle = |0\rangle$
*   $Z|1\rangle = -|1\rangle$

This corresponds to the matrix:
$$ Z = \begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix} $$

#### 3. The Pauli-Y Gate

The `Y` gate can be thought of as a combination of both an `X` and `Z` operation, with a complex phase factor.

*   $Y|0\rangle = i|1\rangle$
*   $Y|1\rangle = -i|0\rangle$

This corresponds to the matrix:
$$ Y = \begin{bmatrix} 0 & -i \\ i & 0 \end{bmatrix} $$

### Outer Products

Another way to construct operators is with the **outer product**. While an inner product $\langle u|v \rangle$ (row × column) gives a scalar, an outer product $|u\rangle\langle v|$ (column × row) produces a matrix.

For example, the operator $A = |0\rangle\langle 1|$ is:
$$ A = \begin{bmatrix} 1 \\ 0 \end{bmatrix} \begin{bmatrix} 0 & 1 \end{bmatrix} = \begin{bmatrix} 0 & 1 \\ 0 & 0 \end{bmatrix} $$
This specific operator transforms any component in the $|1\rangle$ direction into the $|0\rangle$ direction.

# 8. Representing Operators with Matrices

In quantum mechanics, we represent quantum states as **column vectors (kets)**. Similarly, **operators** act as linear transformations on these vectors and can be represented as **matrices**. This reduces the action of an operator on a state to simple matrix-vector multiplication.

$$ |\psi^\prime\rangle = \hat{A}|\psi\rangle $$

An operator $\hat{A}$ in an $n$-dimensional vector space is represented by an $n \times n$ matrix.

---

### Finding Matrix Elements of an Operator

To represent an operator $\hat{A}$ as a matrix, we must know how it acts on the **basis vectors** of the space. If $\{|u_1\rangle, |u_2\rangle, \dots, |u_n\rangle\}$ is a basis for the vector space, then we can compute the matrix elements of $\hat{A}$ using the following formula:

$$ \langle u_i|\hat{A}|u_j\rangle = A_{ij} $$

The resulting matrix $\hat{A}$ would then look like this:

$$ \hat{A} =
\begin{bmatrix}
\langle u_1|\hat{A}|u_1\rangle & \langle u_1|\hat{A}|u_2\rangle & \cdots & \langle u_1|\hat{A}|u_n\rangle \\
\langle u_2|\hat{A}|u_1\rangle & \langle u_2|\hat{A}|u_2\rangle & \cdots & \langle u_2|\hat{A}|u_n\rangle \\
\vdots & \vdots & \ddots & \vdots \\
\langle u_n|\hat{A}|u_1\rangle & \langle u_n|\hat{A}|u_2\rangle & \cdots & \langle u_n|\hat{A}|u_n\rangle
\end{bmatrix}
$$

### Operators in Different Bases

Importantly, the matrix representation of an operator depends on the **basis**. That means the same operator $\hat{A}$ might look very different when expressed in another basis, say $\{|v_1\rangle, |v_2\rangle, \dots, |v_n\rangle\}$. 

To find the matrix representation of $\hat{A}$ in a new basis $\{|v_i\rangle\}$, we compute the matrix elements:

$$ \langle v_i|\hat{A}|v_j\rangle $$

This reflects how much the $j$-th basis vector $|v_j\rangle$ is transformed into the $i$-th basis vector $|v_i\rangle$ under the action of $\hat{A}$.

Note: The elements of the matrix depend on the basis, so matrix representations for the same operator (like $\langle u_i|\hat{A}|u_j\rangle$ and $\langle v_i|\hat{A}|v_j\rangle$) may be different.

# 9. Application: BB84 Quantum Key Distribution

You've now built a solid foundation in the linear algebra of quantum computing. You understand qubits as vectors, measurement as probability, bases and basis changes, and how operators transform states.

We're going to implement the **BB84 Protocol** — the world's first quantum cryptography scheme, invented by Charles Bennett and Gilles Brassard in 1984. This protocol allows two people to create a **perfectly secure secret key** using the quantum properties you've already learned about.

---

## The Problem: Secure Communication

Imagine two friends, **Alice** and **Bob**, who want to communicate secretly. They need a shared secret key to encrypt their messages. But there's a problem: **Eve** (an eavesdropper) is listening on their communication channel!

In classical communication, if Eve intercepts the key, Alice and Bob would never know. But quantum mechanics gives us something amazing: **you cannot measure a quantum state without disturbing it**. This is exactly what we learned earlier — measurement collapses superposition.

---

## The Key Insight: Two Incompatible Bases

Remember our two bases from earlier?

| Basis | States | Names |
|-------|--------|-------|
| **Z (Standard)** | $\|0\rangle = \begin{bmatrix} 1 \\ 0 \end{bmatrix}$, $\|1\rangle = \begin{bmatrix} 0 \\ 1 \end{bmatrix}$ | "computational basis" |
| **X (Hadamard)** | $\|+\rangle = \frac{1}{\sqrt{2}}\begin{bmatrix} 1 \\ 1 \end{bmatrix}$, $\|-\rangle = \frac{1}{\sqrt{2}}\begin{bmatrix} 1 \\ -1 \end{bmatrix}$ | "diagonal basis" |

Here's the crucial fact you proved in Exercise 4:

- If you prepare $|+\rangle$ and measure in the **X basis** → you get `0` with 100% certainty
- If you prepare $|+\rangle$ and measure in the **Z basis** → you get `0` or `1` with 50/50 probability!

**Measuring in the wrong basis destroys the information.** This is the heart of BB84.

---

## How BB84 Works: Step by Step

### Step 1: Alice Prepares Qubits
Alice wants to send a random sequence of bits. For each bit, she:
1. Randomly picks a **basis** (Z or X)
2. Encodes her bit as a qubit in that basis:
   - Z basis: `0` → $|0\rangle$, `1` → $|1\rangle$
   - X basis: `0` → $|+\rangle$, `1` → $|-\rangle$
3. Sends the qubit to Bob

### Step 2: Bob Measures
Bob receives each qubit but **doesn't know which basis Alice used**. So he:
1. Randomly picks a basis (Z or X)
2. Measures the qubit in that basis
3. Records his result

### Step 3: Basis Comparison (Public)
After all qubits are sent and measured:
1. Alice and Bob publicly announce which **bases** they used (NOT the bits!)
2. They **keep** only the bits where they happened to use the **same basis**
3. They **discard** the rest

When their bases matched, Bob's measurement gives the same bit Alice sent — this becomes their **shared secret key**!

### Step 4: Eavesdropper Detection
Here's the magic. If Eve tried to intercept and measure the qubits:
- Eve didn't know the basis either, so she guessed
- When Eve guessed wrong, she **collapsed the qubit to the wrong state**
- Even when Alice and Bob's bases match, Eve's interference causes errors

Alice and Bob sacrifice a portion of their key and compare it publicly. If they see errors, they know Eve was listening!

---

## Why Eve Cannot Win

Let's trace through what happens when Eve intercepts:

1. Alice sends $|+\rangle$ (bit `0` in X basis)
2. Eve intercepts — but she doesn't know the basis!
3. Eve guesses Z basis and measures → gets `0` or `1` randomly (say `0`)
4. The qubit **collapses** to $|0\rangle$ — Eve sends this to Bob
5. Bob picks X basis (matching Alice) and measures $|0\rangle$
6. Bob gets `0` or `1` with 50/50 probability — **even though Alice and Bob used the same basis!**

Eve's measurement corrupted the quantum state. Alice and Bob can detect this corruption.

**The bottom line**: Eve cannot copy a qubit (this is called the "no-cloning theorem"), and she cannot measure without disturbing. Quantum mechanics protects the communication!

```python
def eve_intercept(qubit):
    """
    Eve intercepts a qubit, measures it, and creates a new qubit.
    She doesn't know the correct basis, so she guesses randomly.
    
    Args:
        qubit (np.ndarray): The original qubit state
    
    Returns:
        np.ndarray: The qubit Eve sends to Bob (potentially corrupted!)
    """
    # Eve randomly picks a basis (she's guessing!)
    eve_basis = np.random.choice(['Z', 'X'])
    
    # Eve measures the qubit in her chosen basis
    eve_result = bob_measure(qubit, eve_basis)
    
    # Eve re-encodes what she measured and sends it to Bob
    eve_qubit = alice_encode(eve_result, eve_basis)
    
    return eve_qubit


def bb84_with_eve(n_bits):
    """
    Simulates BB84 with an eavesdropper (Eve) in the middle.
    """
    # Step 1: Alice generates random bits and bases
    alice_bits = np.random.randint(0, 2, n_bits)
    alice_bases = np.random.choice(['Z', 'X'], n_bits)
    
    # Step 2: Bob chooses random bases
    bob_bases = np.random.choice(['Z', 'X'], n_bits)
    
    # Step 3: Alice encodes, EVE INTERCEPTS, Bob measures
    bob_results = []
    for i in range(n_bits):
        # Alice encodes
        qubit = alice_encode(alice_bits[i], alice_bases[i])
        
        # EVE INTERCEPTS AND CORRUPTS THE QUBIT!
        corrupted_qubit = eve_intercept(qubit)
        
        # Bob measures the (potentially corrupted) qubit
        bob_result = bob_measure(corrupted_qubit, bob_bases[i])
        bob_results.append(bob_result)
    
    bob_results = np.array(bob_results)
    
    # Step 4: Basis reconciliation
    matching_bases = alice_bases == bob_bases
    alice_key = alice_bits[matching_bases]
    bob_key = bob_results[matching_bases]
    
    # Step 5: Error detection!
    errors = alice_key != bob_key
    error_rate = np.mean(errors) if len(errors) > 0 else 0
    
    print("=" * 60)
    print("BB84 PROTOCOL WITH EAVESDROPPER (EVE)")
    print("=" * 60)
    print(f"\n Alice's key: {alice_key}")
    print(f"\n Bob's key:   {bob_key}")
    print(f"\n Errors:      {errors.astype(int)}")
    print(f"\n ERROR RATE: {error_rate*100:.1f}%")
    
    if error_rate > 0.1:  # Threshold for suspicion
        print("\n ALERT: High error rate detected!")
        print("   Someone is eavesdropping! Abort and start over!")
    else:
        print("\n Error rate acceptable. Channel appears secure.")
    
    return alice_key, bob_key, error_rate
```
